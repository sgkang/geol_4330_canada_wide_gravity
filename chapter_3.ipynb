```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-1, 1, 20)
z = np.linspace(-1, 1, 20)
X, Z = np.meshgrid(x, z)
C = 1.
r = np.sqrt(X**2 + Z**2)
Bx = -3 * C * X * Z / r**5.
Bz = -3 * C * Z**2 / r**5. + C/r**3
out = plt.pcolormesh(X, Z, Bx)

plt.colorbar(out)
```

```python
Bx_amp = np.sqrt(Bx**2 + Bz**2)
out = plt.quiver(X, Z, Bx/Bx_amp, Bz/Bx_amp)

plt.colorbar(out)
```

```python
import numpy as np
from scipy.special import lpmv
x = np.linspace(-1, 1, 50)
# zonal harmonics
m = 3
for l in range(5):
  plt.plot(x, lpmv(m, l, x), label=f"$P_{l}(x)$")
plt.legend()
plt.xlabel("x")
plt.ylabel("$P_n$(x)")
plt.grid(True, alpha=0.5)
```
