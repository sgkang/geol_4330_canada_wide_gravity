```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-2, 2, 21)
y = np.linspace(-2, 2, 21)
X, Y = np.meshgrid(x, y)
```

```python
def foo(x, y):
  return x*np.exp(-x**2 - y**2)
```

```python
F = X * np.exp(-X**2 - Y**2)
```

```python
out = plt.pcolormesh(X, Y, F)
plt.colorbar(out)
```

```python
Fx = np.exp(-X**2 - Y**2) * (-2*X**2+1)
Fy = np.exp(-X**2 - Y**2) * (-2*X*Y)
```

```python
F_amp = np.sqrt(Fx**2 + Fy**2)
```

```python
Fx_norm = Fx / F_amp
Fy_norm = Fy / F_amp
```

```python
out = plt.pcolormesh(X, Y, F)
plt.colorbar(out)
plt.quiver(X, Y, Fx_norm, Fy_norm, color='white')
```
